```python
import pandas as pd

products = pd.read_csv("products_analysis_ready.csv")
reviews = pd.read_csv("reviews_analysis_ready.csv")

print("Products shape:", products.shape)
print("Reviews shape:", reviews.shape)
```

```python
print("PRODUCT COLUMNS:")
print(products.columns.tolist())

print("\nREVIEW COLUMNS:")
print(reviews.columns.tolist())

print("\nPRODUCT SAMPLE:")
display(products.head())

print("\nREVIEW SAMPLE:")
display(reviews.head())
```

```python
print("========== PRODUCT DATASET ==========")
display(products.describe().T)

print("\n========== REVIEW DATASET ==========")
display(reviews.describe().T)

print("\n========== PRODUCT CATEGORIES ==========")
print(products["primary_category"].value_counts().head(15))

print("\n========== CUSTOMER SKIN TYPES ==========")
print(reviews["skin_type"].value_counts().head(15))

print("\n========== CUSTOMER SKIN TONES ==========")
print(reviews["skin_tone"].value_counts().head(15))
```

```python
import matplotlib.pyplot as plt

# 1. Product category distribution
plt.figure(figsize=(10, 5))
products["primary_category"].value_counts().plot(kind="bar")
plt.title("Product Distribution by Primary Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# 2. Product rating distribution
plt.figure(figsize=(8, 5))
products["rating"].plot(kind="hist", bins=20, edgecolor="black")
plt.title("Distribution of Product Ratings")
plt.xlabel("Product Rating")
plt.ylabel("Number of Products")
plt.tight_layout()
plt.show()


# 3. Customer skin type distribution
plt.figure(figsize=(8, 5))
reviews["skin_type"].value_counts().plot(kind="bar")
plt.title("Reviews by Customer Skin Type")
plt.xlabel("Skin Type")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()


# 4. Customer skin tone distribution
plt.figure(figsize=(10, 5))
reviews["skin_tone"].value_counts().plot(kind="bar")
plt.title("Reviews by Customer Skin Tone")
plt.xlabel("Skin Tone")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()


# 5. Customer recommendation distribution
reviews["is_recommended"].value_counts().sort_index().plot(
    kind="bar", figsize=(7, 5), edgecolor="black"
)
plt.title("Customer Product Recommendation Distribution")
plt.xlabel("Recommendation (0 = No, 1 = Yes)")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
skin_recommendation = reviews.groupby("skin_type").agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).round(3)

display(skin_recommendation)
```

```python
skin_tone_analysis = reviews.groupby("skin_tone").agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).round(3)

skin_tone_analysis = skin_tone_analysis.sort_values(
    "total_reviews", ascending=False
)

display(skin_tone_analysis)
```

```python
category_skin_analysis = reviews.groupby(
    ["skin_type", "product_name"]
).agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).reset_index()

category_skin_analysis = category_skin_analysis.sort_values(
    "total_reviews", ascending=False
)

display(category_skin_analysis.head(20))
```

```python
# Connect product categories to customer reviews

product_categories = products[
    ["product_id", "primary_category"]
].drop_duplicates("product_id")

reviews_with_category = reviews.merge(
    product_categories,
    on="product_id",
    how="left",
    validate="many_to_one"
)

# Compare customer feedback across skin types and categories

category_compatibility = reviews_with_category.groupby(
    ["skin_type", "primary_category"]
).agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).reset_index()

# Keep combinations with at least 100 reviews
category_compatibility = category_compatibility[
    category_compatibility["total_reviews"] >= 100
].copy()

category_compatibility["recommendation_rate"] *= 100

category_compatibility = category_compatibility.round(2)

display(
    category_compatibility.sort_values(
        ["primary_category", "recommendation_rate"],
        ascending=[True, False]
    )
)
```

```python
print("PRODUCT CATEGORY DISTRIBUTION:")
print(products["primary_category"].value_counts())

print("\nREVIEW CATEGORY DISTRIBUTION:")
print(
    reviews_with_category["primary_category"]
    .value_counts(dropna=False)
)

print("\nUNMATCHED REVIEW ROWS:")
print(
    reviews_with_category["primary_category"].isna().sum()
)

print("\nUNIQUE REVIEWED PRODUCTS:")
print(reviews["product_id"].nunique())
```

```python
skin_tone_skincare = reviews_with_category.groupby(
    "skin_tone"
).agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).reset_index()

skin_tone_skincare["recommendation_rate"] *= 100

skin_tone_skincare = skin_tone_skincare[
    skin_tone_skincare["total_reviews"] >= 100
].sort_values("total_reviews", ascending=False).round(2)

display(skin_tone_skincare)
```

```python
import numpy as np

# 1. Rating categories
reviews["rating_category"] = pd.cut(
    reviews["rating"],
    bins=[0, 2, 3, 5],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

# 2. Review text length
reviews["review_text_length"] = (
    reviews["review_text"].fillna("").astype(str).str.len()
)

# 3. Review title length
reviews["review_title_length"] = (
    reviews["review_title"].fillna("").astype(str).str.len()
)

# 4. Positive feedback ratio
reviews["positive_feedback_ratio"] = np.where(
    reviews["total_feedback_count"] > 0,
    reviews["total_pos_feedback_count"] /
    reviews["total_feedback_count"],
    0
)

# 5. Product price groups
reviews["price_category"] = pd.cut(
    reviews["price_usd"],
    bins=[0, 25, 50, 100, np.inf],
    labels=["Budget", "Mid-range", "Premium", "Luxury"],
    include_lowest=True
)

print("Feature engineering completed!")
display(reviews[[
    "rating", "rating_category",
    "review_text_length",
    "review_title_length",
    "positive_feedback_ratio",
    "price_usd", "price_category"
]].head(10))
```

```python
print("FEATURE SUMMARY")
print(reviews[[
    "rating_category",
    "review_text_length",
    "review_title_length",
    "positive_feedback_ratio",
    "price_category"
]].describe(include="all"))

print("\nRATING CATEGORY COUNTS")
print(reviews["rating_category"].value_counts())

print("\nPRICE CATEGORY COUNTS")
print(reviews["price_category"].value_counts())

print("\nMISSING VALUES IN NEW FEATURES")
print(reviews[[
    "rating_category",
    "review_text_length",
    "review_title_length",
    "positive_feedback_ratio",
    "price_category"
]].isna().sum())
```

```python
import matplotlib.pyplot as plt

# Rating category distribution
plt.figure(figsize=(8, 5))
reviews["rating_category"].value_counts().reindex(
    ["Low", "Medium", "High"]
).plot(kind="bar", edgecolor="black")

plt.title("Distribution of Engineered Rating Categories")
plt.xlabel("Rating Category")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# Price category distribution
plt.figure(figsize=(8, 5))
reviews["price_category"].value_counts().reindex(
    ["Budget", "Mid-range", "Premium", "Luxury"]
).plot(kind="bar", edgecolor="black")

plt.title("Review Distribution by Product Price Category")
plt.xlabel("Price Category")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# Review text length distribution
plt.figure(figsize=(8, 5))
reviews["review_text_length"].plot(
    kind="hist", bins=50, edgecolor="black"
)

plt.title("Distribution of Review Text Length")
plt.xlabel("Characters")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

```python
price_rating_analysis = reviews.groupby("price_category").agg(
    total_reviews=("rating", "count"),
    avg_rating=("rating", "mean"),
    recommendation_rate=("is_recommended", "mean")
).round(3)

price_rating_analysis["recommendation_rate"] *= 100

display(price_rating_analysis)
```

```python
from scipy.stats import chi2_contingency

# Remove unknown skin types and missing recommendation labels
test_data = reviews[
    (reviews["skin_type"] != "Unknown") &
    (reviews["is_recommended"].notna())
].copy()

contingency_table = pd.crosstab(
    test_data["skin_type"],
    test_data["is_recommended"]
)

chi2, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("CHI-SQUARE TEST: SKIN TYPE VS RECOMMENDATION")
print("Chi-square statistic:", round(chi2, 3))
print("Degrees of freedom:", dof)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Statistically significant association.")
else:
    print("Result: No statistically significant association.")
```

```python
import numpy as np

n = contingency_table.to_numpy().sum()
min_dimension = min(contingency_table.shape) - 1

cramers_v = np.sqrt(
    chi2 / (n * min_dimension)
)

print("CRAMER'S V:", round(cramers_v, 4))
```

```python
from scipy.stats import chi2_contingency
import numpy as np

# Remove missing recommendation labels
skin_tone_data = reviews[
    reviews["is_recommended"].notna()
].copy()

# Create contingency table
contingency_tone = pd.crosstab(
    skin_tone_data["skin_tone"],
    skin_tone_data["is_recommended"]
)

# Chi-square test
chi2_tone, p_value_tone, dof_tone, expected_tone = (
    chi2_contingency(contingency_tone)
)

# Cramér's V
n = contingency_tone.to_numpy().sum()
min_dimension = min(contingency_tone.shape) - 1

cramers_v_tone = np.sqrt(
    chi2_tone / (n * min_dimension)
)

print("CHI-SQUARE TEST: SKIN TONE VS RECOMMENDATION")
print("Chi-square statistic:", round(chi2_tone, 3))
print("Degrees of freedom:", dof_tone)
print("P-value:", p_value_tone)
print("Cramer's V:", round(cramers_v_tone, 4))

if p_value_tone < 0.05:
    print("Result: Statistically significant association.")
else:
    print("Result: No statistically significant association.")
```
