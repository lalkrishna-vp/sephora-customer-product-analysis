```python
import pandas as pd

file_path = "/content/week4_outputs/reviews_with_engineered_features.csv"

df = pd.read_csv(file_path)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check whether the Week 5 model features exist in our Week 4 dataset

import joblib

lr_model = joblib.load("/content/week5_outputs/logistic_regression_model.pkl")

preprocessor = lr_model.named_steps["preprocessor"]

print("Week 5 numerical features:")
print(preprocessor.transformers_[0][2])

print("\nWeek 5 categorical features:")
print(preprocessor.transformers_[1][2])

print("\nTarget column:")
print("is_recommended")

print("\nDataset shape:")
print(df.shape)
```

```python
# Week 6 - Prepare Evaluation Data Using Exact Week 5 Features

import pandas as pd

# Load the Week 5 model
import joblib

lr_model = joblib.load("/content/week5_outputs/logistic_regression_model.pkl")

# Get the preprocessing pipeline
preprocessor = lr_model.named_steps["preprocessor"]

# Get the exact numerical and categorical features used by Week 5
numerical_features = preprocessor.transformers_[0][2]
categorical_features = preprocessor.transformers_[1][2]

target = "is_recommended"

# Keep only features that actually exist in our dataset
available_numerical = [
    col for col in numerical_features if col in df.columns
]

available_categorical = [
    col for col in categorical_features if col in df.columns
]

evaluation_features = available_numerical + available_categorical

print("Week 6 Evaluation Data")
print("=" * 50)

print("\nNumerical features used:")
print(available_numerical)

print("\nCategorical features used:")
print(available_categorical)

print("\nTarget:")
print(target)

print("\nMissing Week 5 features:")
missing_features = [
    col for col in numerical_features + categorical_features
    if col not in df.columns
]
print(missing_features)

# Prepare X and y
X = df[evaluation_features].copy()
y = df[target].copy()

print("\nEvaluation data prepared")
print("=" * 50)
print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts(dropna=False))

print("\nTarget proportions:")
print(y.value_counts(normalize=True, dropna=False))
```

```python
# Week 6 - Evaluate Week 5 Models

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
import joblib

print("Week 6 Model Evaluation")
print("=" * 50)

# Keep only rows with a known target
known_target_mask = y.notna()

X_eval = X.loc[known_target_mask].copy()
y_eval = y.loc[known_target_mask].copy()

print("Evaluation samples:", len(y_eval))
print("Recommended:", (y_eval == 1).sum())
print("Not recommended:", (y_eval == 0).sum())

# Load Week 5 models
lr_model = joblib.load(
    "/content/week5_outputs/logistic_regression_model.pkl"
)

dt_model = joblib.load(
    "/content/week5_outputs/decision_tree_model.pkl"
)

# Predictions
lr_pred = lr_model.predict(X_eval)
dt_pred = dt_model.predict(X_eval)

# Evaluation function
def evaluate_model(name, y_true, y_pred):
    print(f"\n{name}")
    print("-" * 40)

    print("Accuracy :", round(accuracy_score(y_true, y_pred), 4))
    print("Precision:", round(precision_score(y_true, y_pred), 4))
    print("Recall   :", round(recall_score(y_true, y_pred), 4))
    print("F1-Score :", round(f1_score(y_true, y_pred), 4))

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

# Evaluate both models
evaluate_model("Logistic Regression", y_eval, lr_pred)
evaluate_model("Decision Tree", y_eval, dt_pred)
```

```python
# Week 6 - Model Comparison Visualization

import matplotlib.pyplot as plt
import numpy as np

models = ["Logistic Regression", "Decision Tree"]

accuracy = [
    accuracy_score(y_eval, lr_pred),
    accuracy_score(y_eval, dt_pred)
]

precision = [
    precision_score(y_eval, lr_pred),
    precision_score(y_eval, dt_pred)
]

recall = [
    recall_score(y_eval, lr_pred),
    recall_score(y_eval, dt_pred)
]

f1 = [
    f1_score(y_eval, lr_pred),
    f1_score(y_eval, dt_pred)
]

x = np.arange(len(models))
width = 0.2

plt.figure(figsize=(10, 6))

plt.bar(x - 1.5*width, accuracy, width, label="Accuracy")
plt.bar(x - 0.5*width, precision, width, label="Precision")
plt.bar(x + 0.5*width, recall, width, label="Recall")
plt.bar(x + 1.5*width, f1, width, label="F1-Score")

plt.xticks(x, models)
plt.ylim(0.94, 1.0)
plt.ylabel("Score")
plt.title("Week 6 Model Performance Comparison")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
# Week 6 - Confusion Matrix Comparison

from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_eval,
    lr_pred,
    ax=axes[0],
    values_format="d"
)
axes[0].set_title("Logistic Regression - Confusion Matrix")

ConfusionMatrixDisplay.from_predictions(
    y_eval,
    dt_pred,
    ax=axes[1],
    values_format="d"
)
axes[1].set_title("Decision Tree - Confusion Matrix")

plt.tight_layout()
plt.show()
```

```python
# Week 6 - Business Impact Analysis

import pandas as pd

business_impact = pd.DataFrame({
    "Metric": [
        "Evaluation Samples",
        "Recommended Reviews",
        "Not Recommended Reviews",
        "Logistic Regression Accuracy",
        "Logistic Regression Precision",
        "Logistic Regression Recall",
        "Logistic Regression F1-Score",
        "Decision Tree Accuracy",
        "Decision Tree Precision",
        "Decision Tree Recall",
        "Decision Tree F1-Score"
    ],
    "Value": [
        len(y_eval),
        (y_eval == 1).sum(),
        (y_eval == 0).sum(),
        accuracy_score(y_eval, lr_pred),
        precision_score(y_eval, lr_pred),
        recall_score(y_eval, lr_pred),
        f1_score(y_eval, lr_pred),
        accuracy_score(y_eval, dt_pred),
        precision_score(y_eval, dt_pred),
        recall_score(y_eval, dt_pred),
        f1_score(y_eval, dt_pred)
    ]
})

display(business_impact)

print("\nBusiness Impact Summary")
print("=" * 50)
print("Logistic Regression is the preferred model based on")
print("higher accuracy, recall, and F1-score.")
print("Decision Tree has slightly higher precision.")
```
