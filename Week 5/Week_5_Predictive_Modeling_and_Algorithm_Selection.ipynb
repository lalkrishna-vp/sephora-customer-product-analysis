```python
# ============================================================
# WEEK 5 - LOAD WEEK 4 BACKUP
# ============================================================

import os
import zipfile
import pandas as pd
from google.colab import files

# ------------------------------------------------------------
# 1. Upload the Week 4 ZIP
# ------------------------------------------------------------

print("📦 Select your Week 4 backup ZIP file...")

uploaded = files.upload()

# Get the actual uploaded filename
zip_filename = next(iter(uploaded.keys()))

print(f"\n✅ Uploaded: {zip_filename}")

# ------------------------------------------------------------
# 2. Create extraction folder
# ------------------------------------------------------------

extract_folder = "/content/week4_backup"

os.makedirs(extract_folder, exist_ok=True)

# ------------------------------------------------------------
# 3. Extract EVERYTHING
# ------------------------------------------------------------

with zipfile.ZipFile(zip_filename, "r") as zip_ref:
    zip_ref.extractall(extract_folder)

print("\n✅ Week 4 backup extracted successfully!")

# ------------------------------------------------------------
# 4. List ALL extracted files
# ------------------------------------------------------------

all_files = []

print("\n📁 EXTRACTED FILES")
print("=" * 60)

for root, dirs, files_list in os.walk(extract_folder):
    for file in files_list:
        full_path = os.path.join(root, file)
        all_files.append(full_path)
        print(file)

print(f"\nTotal files extracted: {len(all_files)}")

# ------------------------------------------------------------
# 5. Find CSV files
# ------------------------------------------------------------

csv_files = [
    file for file in all_files
    if file.lower().endswith(".csv")
]

print("\n📊 CSV FILES FOUND")
print("=" * 60)

for file in csv_files:
    print(os.path.basename(file))

print(f"\nTotal CSV files: {len(csv_files)}")

# ------------------------------------------------------------
# 6. Load every CSV
# ------------------------------------------------------------

week4_data = {}

print("\n🔄 LOADING DATASETS")
print("=" * 60)

for file_path in csv_files:

    filename = os.path.basename(file_path)
    key = os.path.splitext(filename)[0]

    try:
        df = pd.read_csv(file_path)

        week4_data[key] = df

        print(f"\n✅ {filename}")
        print(f"   Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")

    except Exception as e:
        print(f"\n❌ Failed to load {filename}")
        print(f"   Error: {e}")

# ------------------------------------------------------------
# 7. Summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("🎯 WEEK 4 DATASETS READY FOR WEEK 5")
print("=" * 60)

for name, df in week4_data.items():
    print(f"{name} → {df.shape}")

# ------------------------------------------------------------
# 8. Load the main Week 4 engineered dataset
# ------------------------------------------------------------

if "reviews_with_engineered_features" in week4_data:

    reviews = week4_data["reviews_with_engineered_features"]

    print("\n" + "=" * 60)
    print("⭐ MAIN DATASET READY")
    print("=" * 60)

    print(f"Shape: {reviews.shape[0]:,} rows × {reviews.shape[1]} columns")

    print("\nColumns:")
    print(reviews.columns.tolist())

    print("\nFirst 5 rows:")
    display(reviews.head())

else:
    print("\n⚠️ Main engineered dataset was not found.")
```

```python
# ============================================================
# WEEK 5 - STEP 1: DEFINE THE PREDICTIVE TARGET
# ============================================================

# Main Week 4 dataset
df = reviews.copy()

print("Dataset shape:", df.shape)

print("\n" + "=" * 60)
print("ALL COLUMNS")
print("=" * 60)

for i, col in enumerate(df.columns, 1):
    print(f"{i:2}. {col}")

# ------------------------------------------------------------
# Check the recommendation target
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TARGET: is_recommended")
print("=" * 60)

print("\nData type:")
print(df["is_recommended"].dtype)

print("\nValue counts:")
print(df["is_recommended"].value_counts(dropna=False))

print("\nPercentage distribution:")
print(
    df["is_recommended"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)

print("\nMissing target values:")
print(df["is_recommended"].isna().sum())
```

```python
# ============================================================
# WEEK 5 - STEP 2: PREPARE MODELING DATA
# ============================================================

# Work only with rows where the target is known
model_df = df[df["is_recommended"].notna()].copy()

print("Original dataset:", df.shape)
print("Labeled dataset:", model_df.shape)

# ------------------------------------------------------------
# Convert target to integer
# ------------------------------------------------------------

model_df["is_recommended"] = model_df["is_recommended"].astype(int)

# ------------------------------------------------------------
# Select initial predictive features
# ------------------------------------------------------------

features = [
    "rating",
    "price_usd",
    "helpfulness",
    "total_feedback_count",
    "total_neg_feedback_count",
    "total_pos_feedback_count",
    "skin_tone",
    "eye_color",
    "skin_type",
    "hair_color",
    "review_text_length",
    "review_title_length",
    "positive_feedback_ratio",
    "price_category",
    "rating_category",
    "review_text_was_missing",
    "review_title_was_missing",
    "helpfulness_was_missing",
    "total_feedback_count_was_missing",
    "total_neg_feedback_count_was_missing",
    "total_pos_feedback_count_was_missing",
    "price_usd_was_missing"
]

# Keep only features that actually exist
features = [col for col in features if col in model_df.columns]

X = model_df[features].copy()
y = model_df["is_recommended"].copy()

print("\n" + "=" * 60)
print("SELECTED FEATURES")
print("=" * 60)

for i, feature in enumerate(features, 1):
    print(f"{i:2}. {feature}")

print("\nNumber of features:", len(features))

print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentages:")
print(y.value_counts(normalize=True).mul(100).round(2))

print("\nMissing values in selected features:")
print(X.isna().sum().sort_values(ascending=False))
```

```python
# ============================================================
# WEEK 5 - STEP 3: TRAIN / TEST SPLIT + PREPROCESSING
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# ------------------------------------------------------------
# 1. Separate numerical and categorical features
# ------------------------------------------------------------

categorical_features = [
    "skin_tone",
    "eye_color",
    "skin_type",
    "hair_color",
    "price_category",
    "rating_category"
]

numerical_features = [
    "rating",
    "price_usd",
    "helpfulness",
    "total_feedback_count",
    "total_neg_feedback_count",
    "total_pos_feedback_count",
    "review_text_length",
    "review_title_length",
    "positive_feedback_ratio",
    "review_text_was_missing",
    "review_title_was_missing",
    "helpfulness_was_missing",
    "total_feedback_count_was_missing",
    "total_neg_feedback_count_was_missing",
    "total_pos_feedback_count_was_missing",
    "price_usd_was_missing"
]

# ------------------------------------------------------------
# 2. Verify selected columns
# ------------------------------------------------------------

print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))
print("Total features:", len(categorical_features) + len(numerical_features))

# ------------------------------------------------------------
# 3. Create X and y
# ------------------------------------------------------------

X = model_df[categorical_features + numerical_features]
y = model_df["is_recommended"]

# ------------------------------------------------------------
# 4. Stratified train/test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 60)
print("TRAIN / TEST SPLIT")
print("=" * 60)

print(f"Training samples: {len(X_train):,}")
print(f"Testing samples:  {len(X_test):,}")

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).mul(100).round(2))

# ------------------------------------------------------------
# 5. Preprocessing pipelines
# ------------------------------------------------------------

# Numerical:
# - Impute missing values using median
# - Standardize numerical features

numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical:
# - Impute missing values using most frequent value
# - Convert categories into numerical one-hot features

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=True
        ))
    ]
)

# ------------------------------------------------------------
# 6. Combine preprocessing
# ------------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("\n" + "=" * 60)
print("PREPROCESSING READY")
print("=" * 60)

print("✅ Numerical features → imputation + scaling")
print("✅ Categorical features → imputation + one-hot encoding")
print("✅ Unknown categories → safely ignored")
print("✅ Stratified split → class proportions preserved")
```

```python
# ============================================================
# WEEK 5 - STEP 4: TRAIN LOGISTIC REGRESSION
# ============================================================

from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Create Logistic Regression pipeline
# ------------------------------------------------------------

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        ))
    ]
)

# ------------------------------------------------------------
# 2. Train the model
# ------------------------------------------------------------

print("🚀 Training Logistic Regression...")
print("Training samples:", f"{len(X_train):,}")

logistic_model.fit(X_train, y_train)

print("\n✅ Logistic Regression training completed!")

# ------------------------------------------------------------
# 3. Predict on test data
# ------------------------------------------------------------

y_pred_lr = logistic_model.predict(X_test)

# ------------------------------------------------------------
# 4. Calculate evaluation metrics
# ------------------------------------------------------------

accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr)
recall_lr = recall_score(y_test, y_pred_lr)
f1_lr = f1_score(y_test, y_pred_lr)

# ------------------------------------------------------------
# 5. Display results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("LOGISTIC REGRESSION RESULTS")
print("=" * 60)

print(f"Accuracy :  {accuracy_lr:.4f}")
print(f"Precision:  {precision_lr:.4f}")
print(f"Recall   :  {recall_lr:.4f}")
print(f"F1-score :  {f1_lr:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

```python
# ============================================================
# WEEK 5 - STEP 5: SAVE LOGISTIC REGRESSION MODEL
# ============================================================

import joblib

joblib.dump(
    logistic_model,
    "/content/logistic_regression_model.pkl"
)

print("✅ Logistic Regression model saved successfully!")
print("File: /content/logistic_regression_model.pkl")
```

```python
# ============================================================
# WEEK 5 - STEP 6: TRAIN DECISION TREE
# ============================================================

from sklearn.tree import DecisionTreeClassifier

# ------------------------------------------------------------
# 1. Create Decision Tree pipeline
# ------------------------------------------------------------

decision_tree_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(
            max_depth=10,
            min_samples_split=20,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=42
        ))
    ]
)

# ------------------------------------------------------------
# 2. Train the model
# ------------------------------------------------------------

print("🌳 Training Decision Tree...")
print("Training samples:", f"{len(X_train):,}")

decision_tree_model.fit(X_train, y_train)

print("\n✅ Decision Tree training completed!")

# ------------------------------------------------------------
# 3. Make predictions
# ------------------------------------------------------------

y_pred_dt = decision_tree_model.predict(X_test)

# ------------------------------------------------------------
# 4. Calculate evaluation metrics
# ------------------------------------------------------------

accuracy_dt = accuracy_score(y_test, y_pred_dt)
precision_dt = precision_score(y_test, y_pred_dt)
recall_dt = recall_score(y_test, y_pred_dt)
f1_dt = f1_score(y_test, y_pred_dt)

# ------------------------------------------------------------
# 5. Display results
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("DECISION TREE RESULTS")
print("=" * 60)

print(f"Accuracy :  {accuracy_dt:.4f}")
print(f"Precision:  {precision_dt:.4f}")
print(f"Recall   :  {recall_dt:.4f}")
print(f"F1-score :  {f1_dt:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
```

```python
# ============================================================
# WEEK 5 - STEP 7: SAVE DECISION TREE MODEL
# ============================================================

import joblib

joblib.dump(
    decision_tree_model,
    "/content/decision_tree_model.pkl"
)

print("✅ Decision Tree model saved successfully!")
print("File: /content/decision_tree_model.pkl")
```

```python
# ============================================================
# WEEK 5 - STEP 8: MODEL COMPARISON
# ============================================================

import pandas as pd

model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree"
    ],
    "Accuracy": [
        accuracy_lr,
        accuracy_dt
    ],
    "Precision": [
        precision_lr,
        precision_dt
    ],
    "Recall": [
        recall_lr,
        recall_dt
    ],
    "F1-Score": [
        f1_lr,
        f1_dt
    ]
})

print("=" * 70)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 70)

print(
    model_comparison.round(4).to_string(index=False)
)

# Identify best model based on F1-score
best_model = model_comparison.loc[
    model_comparison["F1-Score"].idxmax(),
    "Model"
]

print("\n🏆 Best model based on F1-score:", best_model)
```

```python
# ============================================================
# WEEK 5 - STEP 9: LOGISTIC REGRESSION FEATURE IMPORTANCE
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# Get the trained preprocessing step
# ------------------------------------------------------------

fitted_preprocessor = logistic_model.named_steps["preprocessor"]

# Get transformed feature names
feature_names = fitted_preprocessor.get_feature_names_out()

# Get logistic regression coefficients
coefficients = logistic_model.named_steps["classifier"].coef_[0]

# Create feature importance table
feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients,
    "Absolute_Importance": np.abs(coefficients)
})

# Sort by strongest influence
feature_importance = feature_importance.sort_values(
    "Absolute_Importance",
    ascending=False
).reset_index(drop=True)

# ------------------------------------------------------------
# Display top 20 features
# ------------------------------------------------------------

print("=" * 70)
print("TOP 20 FEATURES - LOGISTIC REGRESSION")
print("=" * 70)

print(
    feature_importance.head(20).to_string(index=False)
)

# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

feature_importance.to_csv(
    "/content/logistic_regression_feature_importance.csv",
    index=False
)

print("\n✅ Feature importance saved!")
```

```python
# ============================================================
# WEEK 5 - STEP 10: CROSS-VALIDATION FOR MODEL ROBUSTNESS
# ============================================================

from sklearn.model_selection import StratifiedKFold, cross_validate

# ------------------------------------------------------------
# 1. Define stratified cross-validation
# ------------------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# ------------------------------------------------------------
# 2. Evaluation metrics
# ------------------------------------------------------------

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1"
}

# ------------------------------------------------------------
# 3. Run cross-validation
# ------------------------------------------------------------

print("=" * 70)
print("5-FOLD CROSS-VALIDATION - LOGISTIC REGRESSION")
print("=" * 70)

cv_results = cross_validate(
    logistic_model,
    X,
    y,
    cv=cv,
    scoring=scoring,
    n_jobs=-1
)

# ------------------------------------------------------------
# 4. Display fold results
# ------------------------------------------------------------

for metric in scoring.keys():
    scores = cv_results[f"test_{metric}"]

    print(f"\n{metric.upper()}:")
    print("Fold scores:", np.round(scores, 4))
    print(f"Mean: {scores.mean():.4f}")
    print(f"Std : {scores.std():.4f}")

# ------------------------------------------------------------
# 5. Overall robustness summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CROSS-VALIDATION SUMMARY")
print("=" * 70)

print(f"Mean Accuracy : {cv_results['test_accuracy'].mean():.4f}")
print(f"Mean Precision: {cv_results['test_precision'].mean():.4f}")
print(f"Mean Recall   : {cv_results['test_recall'].mean():.4f}")
print(f"Mean F1-score : {cv_results['test_f1'].mean():.4f}")

print("\n✅ 5-fold cross-validation completed.")
```

```python
# ============================================================
# WEEK 5 - STEP 11: FINAL MODELING RESULTS
# ============================================================

final_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree"
    ],
    "Test Accuracy": [
        accuracy_lr,
        accuracy_dt
    ],
    "Test Precision": [
        precision_lr,
        precision_dt
    ],
    "Test Recall": [
        recall_lr,
        recall_dt
    ],
    "Test F1-Score": [
        f1_lr,
        f1_dt
    ]
})

print("=" * 80)
print("FINAL MODEL PERFORMANCE")
print("=" * 80)

print(final_results.round(4).to_string(index=False))

print("\n" + "=" * 80)
print("SELECTED MODEL")
print("=" * 80)

print("Logistic Regression")
print(f"Cross-validation F1-score: {cv_results['test_f1'].mean():.4f}")
print(f"Cross-validation std:       {cv_results['test_f1'].std():.4f}")

# Save final comparison
final_results.to_csv(
    "/content/week5_model_comparison.csv",
    index=False
)

print("\n✅ Final model comparison saved!")
```
